## 1. Setup

Run this cell once before creating the plots.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
np.random.seed(7)
```

## 2. Histograms

A **histogram** shows how often numeric values fall into different ranges.

Here we simulate PM2.5 concentration measurements.


```python
pm25 = np.random.gamma(shape=4, scale=4, size=200) + 5

plt.figure(figsize=(8, 4))
plt.hist(pm25, bins=15, edgecolor="black")

plt.title("Distribution of PM2.5 Concentration")
plt.xlabel("PM2.5 (µg/m³)")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

A histogram helps answer questions such as:
- Where are most values located?
- How spread out are the values?
- Are there unusually high or low values?


## 3. Comparing groups with boxplots

A **boxplot** gives a compact summary of a numeric variable and is useful for comparing groups.

The example below compares PM10 concentrations at three station types.


```python
station_df = pd.DataFrame({
    "station": np.repeat(["Urban", "Suburban", "Rural"], 60),
    "pm10": np.concatenate([
        np.random.normal(42, 8, 60),
        np.random.normal(34, 6, 60),
        np.random.normal(26, 5, 60)
    ])
})

plt.figure(figsize=(8, 4))
sns.boxplot(data=station_df, x="station", y="pm10")

plt.title("PM10 Distribution by Station Type")
plt.xlabel("Station type")
plt.ylabel("PM10 (µg/m³)")

plt.tight_layout()
plt.show()
```

Use a boxplot when you want to compare the **typical values and spread** of several groups.

In this example, the urban station has higher PM10 values than the suburban and rural stations.


## 4. Relationships with scatter plots

A **scatter plot** shows the relationship between two numeric variables.

Here we compare traffic intensity and PM2.5 concentration.


```python
traffic = np.random.uniform(50, 300, 100)
pm25_traffic = 10 + 0.08 * traffic + np.random.normal(0, 4, 100)

traffic_df = pd.DataFrame({
    "traffic": traffic,
    "pm25": pm25_traffic
})

plt.figure(figsize=(8, 4))
sns.scatterplot(data=traffic_df, x="traffic", y="pm25")

plt.title("Traffic Intensity and PM2.5 Concentration")
plt.xlabel("Traffic intensity (vehicles/hour)")
plt.ylabel("PM2.5 (µg/m³)")

plt.tight_layout()
plt.show()
```

The points suggest a **positive relationship**: higher traffic intensity is generally associated with higher PM2.5 values.

A scatter plot shows association, but it does not prove that one variable causes the other.


## 5. Correlation heatmap

**Correlation** describes the strength and direction of a linear relationship between numeric variables.

Values close to:
- `1` indicate a strong positive relationship,
- `-1` indicate a strong negative relationship,
- `0` indicate little or no linear relationship.


```python
env_df = pd.DataFrame({
    "pm25": np.random.normal(25, 6, 120),
    "temperature": np.random.normal(17, 7, 120),
    "humidity": np.random.normal(68, 10, 120)
})

env_df["pm10"] = env_df["pm25"] * 1.2 + np.random.normal(0, 4, 120)

correlation = env_df.corr()

plt.figure(figsize=(6, 4))
sns.heatmap(correlation, annot=True, fmt=".2f")

plt.title("Correlation Between Environmental Variables")
plt.tight_layout()
plt.show()
```

The heatmap makes it easy to compare several correlations at once.  
Here, PM2.5 and PM10 have the strongest positive relationship.

Remember: **correlation does not imply causation**.
