## 1. Setup

Run the following cell once before creating the plots.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

## 2. Line plots

Use a **line plot** when the x-axis has a natural order, especially time.  
The example below shows average monthly temperature.


```python
months = pd.date_range("2025-01-01", periods=12, freq="MS")
temperature = [2.1, 4.0, 8.7, 13.4, 18.2, 22.1,
               24.7, 24.1, 19.2, 13.0, 7.2, 3.1]

plt.figure(figsize=(9, 4))
plt.plot(months, temperature, marker="o")

plt.title("Average Monthly Air Temperature")
plt.xlabel("Month")
plt.ylabel("Temperature (°C)")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

The plot shows a clear seasonal pattern: temperatures rise toward summer and fall again toward winter.

The most important parts of the code are:
- `plt.plot()` creates the line,
- `plt.title()` adds a title,
- `plt.xlabel()` and `plt.ylabel()` label the axes,
- `plt.show()` displays the figure.


## 3. Bar plots

Use a **bar plot** to compare values across categories.  
Here we compare average NO$_2$ concentration at four monitoring stations.


```python
stations = ["Station A", "Station B", "Station C", "Station D"]
no2 = [18.2, 24.6, 20.1, 27.4]

plt.figure(figsize=(8, 4))
plt.bar(stations, no2)

plt.title("Average NO$_2$ Concentration by Station")
plt.xlabel("Monitoring station")
plt.ylabel("NO$_2$ (µg/m³)")

plt.tight_layout()
plt.show()
```

A simple rule:

- **Line plot:** use it for time or another ordered sequence.
- **Bar plot:** use it to compare categories.


## 4. Plotting a DataFrame with Seaborn

Real datasets are usually stored in tables. A pandas DataFrame keeps related variables together.


```python
air_quality = pd.DataFrame({
    "month": months,
    "temperature": temperature,
    "pm25": [31, 28, 24, 19, 15, 12, 11, 13, 16, 22, 27, 30]
})

air_quality.head()
```

Seaborn works directly with DataFrame column names. This can make plotting code easier to read.


```python
plt.figure(figsize=(9, 4))
sns.lineplot(data=air_quality, x="month", y="pm25", marker="o")

plt.title("Monthly PM2.5 Concentration")
plt.xlabel("Month")
plt.ylabel("PM2.5 (µg/m³)")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

Matplotlib and Seaborn are often used together: Seaborn creates the plot, while Matplotlib functions are used for titles, labels, and figure settings.
